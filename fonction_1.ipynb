```python
# Exemples
# Fonction qui prend deux entrées (deux nombres réelles) et retourne la somme

def sum(a, b):
    d = a + b
    return d
```

```python
r = sum(4, 5)
print(r)
```

**Fonction de type HINTS** 

```python
def sum_type_hint(a: int, b: float) -> float:
    """Somme de deux nombres reçus en entrée.
    """
    d = a + b
    return d
```

```python
r = sum_type_hint(4, 5.0)
r
```

```python
# Fonction de type Hints avec valeurs par défauts
def sum_type_hint_default(a: int = 2, b: float = 2.0) -> float:
    """Somme de deux nombres reçus en entrée.
    """
    d = a + b
    return d
```

```python
sum_type_hint_default(5, 9.0)
```

**Passage des arguments**

```python
# Fonction de type Hints avec valeurs par défauts
def sum_type_hint_default_1(a: int = 2, b: float = 2.0, c: int = 5, d: int = 8) -> float:
    """Somme de tous les nombres reçus en entrée.
    """
    d = a + b + c + d
    return d
```

```python
# Argument positionels
sum_type_hint_default_1(8, 5, 1, 0)

# Passage par variable 
sum_type_hint_default_1(d = 8, b = 5, a = 1, c = 0)
```

***Avant d'exécuter les cellules suivantes, appuyez sur 'Restart' pour redémarrer le kernel (noyau). Sinon, vous aurez une erreur, car la fonction 'sum()' est aussi une fonction définie par défaut dans Python.***

```python
def process_data_stream(*args: int, **kwargs: str) -> dict:
    """
    - *args: Captures an arbitrary number of positional arguments into a tuple.
    - **kwargs: Captures an arbitrary number of keyword arguments into a dictionary.
    """
    total = sum(args)
    metadata = {key: value.upper() for key, value in kwargs.items()}
    return {"sum": total, "metadata": metadata}
```

```python
result = process_data_stream(10, 20, 30, environment="production", region="eu-west-1")
print(f"Result: {result}\n")
```
