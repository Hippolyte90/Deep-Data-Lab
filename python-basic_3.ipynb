## 1. `if / else` — la météo du matin

> **Analogie :** s'il pleut, je prends un parapluie ; sinon, je prends mes lunettes de soleil.

Une condition, c'est une question fermée (oui / non). Selon la réponse, on prend un chemin ou un autre.

```python
il_pleut = False

if il_pleut:
    print("Je prends un parapluie 🌂")
    print("Dans les deux cas : je sors.")

else:
    print("Je prends mes lunettes de soleil 😎")
```

**À toi :** change `il_pleut = True` en `False` et ré-exécute la cellule.

## 2. `if / elif / else` — le feu tricolore

> **Analogie :** rouge = j'arrête, orange = je ralentis, vert = j'avance.

Dès qu'une réponse est « oui », Python arrête de tester les suivantes.

```python
feu = "orange"

if feu == "rouge":
    print("Je m'arrête complètement 🛑")
elif feu == "orange":
    print("Je ralentis, je me prépare 🟠")
else:
    print("J'avance ➡️")
```

**À toi :** essaie `"rouge"`, `"vert"`, puis `"bleu"`. Que se passe-t-il pour une valeur non prévue ?

## 3. `try / except` — le filet du trapéziste

> **Analogie :** le trapéziste tente son saut (`try`). S'il tombe, le filet le rattrape (`except`). Le spectacle continue.

Sans `try / except`, une division par zéro fait planter tout le programme.

```python
# Sans filet — ce code plante
resultat = 10 / 0
```

```python
# Avec filet — le programme survit
try:
    resultat = 10 / 0
    print(f"Résultat : {resultat}")
except ZeroDivisionError:
    print("⚠️ Impossible de diviser par zéro — j'utilise 0 à la place.")
    resultat = 0

print(f"On continue avec resultat = {resultat}")
```

**Autre exemple** — l'utilisateur tape n'importe quoi :

```python
try:
    age = int(input("Ton âge : "))
    print(f"Tu as {age} ans.")
except ValueError:
    print("⚠️ Ce n'est pas un nombre valide.")
```

## 4. Boucle `for` — distribuer les cartes

> **Analogie :** pour chaque joueur autour de la table, je donne une carte. Un joueur = un tour de boucle.

On sait à l'avance combien de tours : autant qu'il y a d'éléments.

```python
joueurs = ["Alice", "Bob", "Claire", "David"]

for joueur in joueurs:
    print(f"{joueur} reçoit 1 carte 🃏")

print("Tout le monde est servi.")
```

**Variante** — répéter un nombre de fois précis avec `range` :

```python
for tour in range(1, 6):   # de 1 à 5 inclus
    print(f"Tour n°{tour}")
```

## 5. Boucle `while` — remplir un verre

> **Analogie :** tant que le verre n'est pas plein, je verse. Je ne sais pas combien de fois — ça dépend de sa taille.

⚠️ Attention à la boucle infinie : quelque chose doit changer à chaque tour pour que la condition finisse par devenir fausse.

```python
niveau = 0

while niveau < 100:
    niveau = niveau + 10
    print(f"Niveau : {niveau}%")

print("Le verre est plein ! 🥛")
```

**Cas concret** — redemander tant que l'utilisateur se trompe :

```python
mot_de_passe = ""

while mot_de_passe != "python123":
    mot_de_passe = input("Mot de passe : ")

print("Bienvenue ! ✅")
```

## 6. `for` vs `while` — lequel choisir ?

| Situation | Choix |
|---|---|
| Tu connais le nombre de tours à l'avance | `for` |
| Tu attends un événement (saisie, seuil, réponse serveur…) | `while` |

**Même exemple, deux approches :**

```python
# Avec for — je connais la fin (5 tours)
print("Version FOR :")
for i in range(5):
    print(f"  Tour {i + 1}")
```

```python
# Avec while — je m'arrête quand une condition change
print("Version WHILE :")
i = 0
while i < 5:
    print(f"  Tour {i + 1}")
    i = i + 1
```

## 7. `break` et `continue` — cinéma et playlist

> **`break` = alerte incendie au cinéma :** tu quittes la salle immédiatement, sans finir le film.
> **`continue` = chanson qu'on saute :** on abandonne ce morceau, mais la playlist continue.

```python
# break — arrêter tout dès qu'on trouve
nombres = [3, 7, 12, 25, 42, 99]

for n in nombres:
    if n > 20:
        print(f"Premier nombre > 20 trouvé : {n}")
        break
    print(f"  Je regarde {n}...")
```

```python
# continue — sauter les nombres pairs
for n in range(1, 11):
    if n % 2 == 0:
        continue          # on saute le reste et on passe au tour suivant
    print(f"Nombre impair : {n}")
```

## 8. Boucles imbriquées — le calendrier

> **Analogie :** pour chaque mois de l'année (boucle extérieure), on parcourt tous ses jours (boucle intérieure).

Attention à la complexité : 12 mois × 30 jours = 360 tours.

```python
mois = ["Janvier", "Février", "Mars"]     # extrait pour l'exemple
jours = [1, 2, 3, 4, 5]                    # 5 premiers jours seulement

for m in mois:
    print(f"--- {m} ---")
    for j in jours:
        print(f"  {m} {j}")
```

## 9. `with` — la porte qui se referme toute seule

> **Analogie :** tu entres, tu fais ce que tu as à faire, la porte se referme derrière toi automatiquement.

`with` garantit que la ressource (fichier, connexion…) sera libérée à la fin — même en cas d'erreur.

```python
# Écrire dans un fichier
with open("mon_fichier_1.txt", "w", encoding="utf-8") as f:
    f.write("Bonjour depuis Python !\n")
    f.write("Ligne 2\n")
# Ici, le fichier est déjà refermé automatiquement.

print("Fichier écrit et refermé.")
```

```python
# Relire le fichier
with open("mon_fichier_1.txt", "r", encoding="utf-8") as f:
    contenu = f.read()

print(contenu)
```

**Sans `with`** il aurait fallu appeler `f.close()` à la main — et un `close` oublié après une erreur = ressource perdue.

## 11. Exercice final — tout combiné 🎯

**Ton mission :** un mini-programme de **suivi des températures de la semaine**.

Le programme doit :
- Demander la température max de chaque jour (`for`)
- Redemander tant que l'entrée n'est pas un nombre valide (`while` + `try / except`)
- Laisser l'utilisateur **sauter un jour** (`continue`) ou **arrêter en cours** (`break`)
- Classer chaque journée en froid / tempéré / chaud (`if / elif / else`)
- Sauvegarder le récap dans un fichier (`with`)

**Essaie d'abord toi-même ↓**, la solution est en dessous.

### Solution commentée

Chaque commentaire `# →` pointe vers un concept vu dans le PowerPoint.

```python
jours = ["Lundi", "Mardi", "Mercredi", "Jeudi", "Vendredi", "Samedi", "Dimanche"]
temperatures = {}

print("🌡️  Enregistre les températures max de la semaine.")
print("   (tape 'skip' pour sauter un jour, 'stop' pour arrêter)\n")

# → for : on parcourt une collection connue (7 jours)
for jour in jours:

    # → while : on ne sait pas combien de fois l'utilisateur va se tromper
    while True:
        reponse = input(f"Température max {jour} (°C) : ").strip().lower()

        # → if / elif / else : on choisit parmi plusieurs cas
        if reponse == "stop":
            break                       # → break : sort de la boucle while
        elif reponse == "skip":
            reponse = None
            break
        else:
            # → try / except : la conversion en nombre peut échouer
            try:
                temperatures[jour] = float(reponse)
                break                   # entrée valide → on sort de la boucle
            except ValueError:
                print("   ⚠️  Ce n'est pas un nombre. Réessaie.")

    # Après la boucle while, on décide de la suite pour la boucle for
    if reponse == "stop":
        print("→ Arrêt anticipé.\n")
        break                           # → break : sort de la boucle for
    if reponse is None:
        print(f"   {jour} sauté.\n")
        continue                        # → continue : passe au jour suivant

# --- Analyse ---
print("\n📊 Analyse de la semaine :\n")

# → for + if / elif / else : on classe chaque journée
for jour, temp in temperatures.items():
    if temp < 10:
        etat = "❄️  Froid"
    elif temp < 20:
        etat = "🌤️  Tempéré"
    else:
        etat = "☀️  Chaud"
    print(f"   {jour:10s} : {temp:5.1f}°C  →  {etat}")

# --- Sauvegarde ---
# → with : le fichier se referme tout seul, même en cas d'erreur
with open("temperatures.txt", "w", encoding="utf-8") as f:
    f.write("Températures de la semaine\n")
    f.write("=" * 30 + "\n")
    for jour, temp in temperatures.items():
        f.write(f"{jour}: {temp}°C\n")

print(f"\n✅ Résultats enregistrés dans temperatures.txt ({len(temperatures)} jour(s))")
```

### Récap des 6 briques utilisées

| Concept | Où, dans l'exercice ? |
|---|---|
| `for` | Parcours des 7 jours, puis parcours du dictionnaire pour l'analyse |
| `while` | Redemande une saisie tant qu'elle est invalide |
| `try / except` | Convertit la saisie en nombre sans faire planter le programme |
| `if / elif / else` | Gère `stop` / `skip` / valeur numérique, puis classe la température |
| `break` | Sort de la boucle de saisie **et** de la boucle des jours |
| `continue` | Passe au jour suivant sans enregistrer |
| `with` | Écrit le récap dans un fichier, refermé automatiquement |
