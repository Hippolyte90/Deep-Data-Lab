## Opérateurs de comparaison

*   ==
*   !=
*   <= or >=
*   <  or >

```python
# Exemple d'utilisation des opérateurs de comparaison en Python

# On définit deux variables
a = 10
b = 20

print("a =", a)
print("b =", b)

# == (égalité)
print("a == b :", a == b)   # False car 10 n'est pas égal à 20

# != (différent)
print("a != b :", a != b)   # True car 10 est différent de 20

# <= (inférieur ou égal)
print("a <= b :", a <= b)   # True car 10 est bien inférieur à 20

# >= (supérieur ou égal)
print("a >= b :", a >= b)   # False car 10 n'est pas supérieur ou égal à 20

# < (strictement inférieur)
print("a < b  :", a < b)    # True car 10 < 20

# > (strictement supérieur)
print("a > b  :", a > b)    # False car 10 n'est pas supérieur à 20
```

## Opérateurs booléens

* and
* or
* not

```python
# Exemple d'utilisation des opérateurs logiques en Python

x = True
y = False

print("x =", x)
print("y =", y)

# AND (et logique)
print("x and y :", x and y)   # False car True ET False = False
print("x and x :", x and x)   # True car True ET True = True

# OR (ou logique)
print("x or y  :", x or y)    # True car True OU False = True
print("y or y  :", y or y)    # False car False OU False = False

# NOT (négation logique)
print("not x   :", not x)     # False car NOT True = False
print("not y   :", not y)     # True car NOT False = True
```
